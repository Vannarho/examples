# Example 1 hello: installed wheel import

This notebook executes against the installed `vannarho-risk-engine` wheel only.

```python
from pathlib import Path
import importlib.util

HERE = Path.cwd().resolve()
RUNTIME_PATH = HERE.parent / "_wheel_runtime.py"
spec = importlib.util.spec_from_file_location("vre_wheel_runtime", RUNTIME_PATH)
if spec is None or spec.loader is None:
    raise RuntimeError(f"Could not load wheel runtime helper from {RUNTIME_PATH}")
runtime = importlib.util.module_from_spec(spec)
spec.loader.exec_module(runtime)

VRE = runtime._import_vre_runtime()
print("hello", VRE.__name__)
```
