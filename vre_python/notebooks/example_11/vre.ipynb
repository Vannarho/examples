# Example 11: EQ/COM SA-CVA smoke runner

This notebook executes one canonical case against the installed `vannarho-risk-engine` wheel only. The aggregate script runner covers the full multi-case manifest.

```python
from pathlib import Path
import os
import subprocess
import sys

HERE = Path.cwd()

def run_script(script_name: str) -> None:
    env = os.environ.copy()
    for key in ("PYTHONPATH", "VRE_PYBIND_FORCE_BUILD_DIR", "VRE_PYBIND_BUILD_DIR", "VRE_NOTEBOOK_BUILD_ROOT"):
        env.pop(key, None)
    subprocess.run([sys.executable, str(HERE / script_name)], cwd=HERE, env=env, check=True)
```

```python
run_script("run_sacva_eqcom.py")
```
